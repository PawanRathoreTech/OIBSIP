```python
category_sales = df.groupby("Product Category")["Total Amount"].sum().sort_values(ascending=False)

print("category-wise Revenue:")
print(category_sales)
```

```python
plt.figure(figsize = (8,4))

category_sales.plot(kind="bar")
plt.title("Total Revenue by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Total revenue")
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```

```python
Gender_count = df["Gender"].value_counts()
print("Gender-wise Transaction Count:")
print(Gender_count)
```

```python
plt.figure(figsize=(8,3))
Gender_count.plot(kind = "bar")
plt.title("Gender-wise Transaction Distrubution")
plt.xlabel("Gender")
plt.ylabel("Number of Transactions")
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```

```python
----Markdown---

### Observation: Gender-wise Transaction Distribution

- Female customers account for 510 transactions (51%).
- Male customers account for 490 transactions (49%).
- The transaction distribution is nearly balanced, with female customers contributing slightly more transactions.
```

```python
Gender_revenue = df.groupby("Gender")["Total Amount"].sum().sort_values(ascending=False)
print("Gender-wise Revenue:")
print(Gender_revenue)
```

```python
---------Markdown-------
### Observation: Revenue by Gender

- Female custome rs generated a total revenue of 232,840.
- Male customers generated a total revenue of 223,160.
- Female customers contributed 9,680 more in revenue than male customers.
- Revenue contribution from both gender groups was relatively balanced.
```

```python
df["Age Group"] = pd.cut(df["Age"],
bins=[17,25,35,45,55,65],
labels=["18-25","26-35","36-45","46-55","56-64"])

age_group_count = df["Age Group"].value_counts().sort_index()

print("Transactions be age group:")
print(age_group_count)
```

```python
plt.figure(figsize=(9,5))

age_group_count.plot(kind="bar")
plt.title("Transaction Distrubution by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Number of Transactions")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
------Markdown---
### Observation: Customer Age Group Distribution

- The 46-55 age group recorded the highest number of transactions (229).
- The 18-25 age group recorded the lowest number of transactions (169).
- Customers across all age groups contributed to sales.
- The transaction distribution is relatively balanced across different age groups.
```

```python
age_group_revenue = df.groupby("Age Group",observed=False)["Total Amount"].sum().sort_values(ascending=False)

print("Revenue by age-group")
print(age_group_revenue)
```

```python
plt.figure(figsize=(9,4))

age_group_revenue.plot(kind="bar")
plt.title("Total Revenue by age group")
plt.xlabel("Age Group")
plt.ylabel("Total Revenue")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
df["month"] = df["Date"].dt.to_period("M")

monthly_sales = df.groupby("month")["Total Amount"].sum()
print(monthly_sales)
```

```python
plt.figure(figsize=(9,5))

monthly_sales.plot(kind="line",marker="o")
plt.title("Monthly sales trend")
plt.xlabel("Month")
plt.xlabel("Total sales revenue")
plt.xticks(rotation = 0)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
----------markdown--------------
### Monthly Sales Trend — Observations

- May 2023 recorded the highest monthly revenue of 53,150.
- September 2023 recorded the lowest monthly revenue of 23,620.
- Sales revenue fluctuated throughout 2023, with no consistent upward or downward trend.
- Revenue increased significantly from September to October.
- January 2024 shows revenue of only 1,530. Since the dataset contains very limited data for January 2024, this value should not be compared directly with full months.
```

```python
category_avg_amount = df.groupby("Product Category")["Total Amount"].mean().sort_values(ascending = False)
print(category_avg_amount)
```

```python
plt.figure(figsize=(8,3))

category_avg_amount.plot(kind="bar",color="skyblue")
plt.title("Average Transaction Amount by product")
plt.xlabel("Product Category")
plt.ylabel("Avg Transaction Amount (₹)")
plt.xticks(rotation = 0)
plt.tight_layout()
plt.show()
```

```python
--------------Markdown----------------

### Observation

* Beauty has the highest average transaction amount at approximately ₹467.48.
* Electronics follows with an average transaction amount of ₹458.79.
* Clothing has the lowest average transaction amount at approximately ₹443.25.
* The difference between the highest and lowest category averages is about ₹24.23, indicating that average transaction amounts are relatively similar across categories.
```

```python
-----------Markdown-----------
## Conclusion

The retail sales dataset contains 1,000 transactions. The analysis examined sales trends, customer demographics, product category performance, and relationships between numerical variables.

The results show that:

* Beauty generated the highest average transaction amount (approximately ₹467.48).
* Clothing recorded the highest total quantity sold (894 units).
* Female customers contributed ₹232,840 in revenue, compared with ₹223,160 from male customers.
* The 46–55 age group generated the highest revenue (₹100,690).
* Monthly and quarterly sales varied over time. The January 2024 data represents only a partial period, so it should not be compared directly with complete months or quarters.
* Price per Unit had a strong positive correlation with Total Amount, while Age showed almost no linear correlation with Total Amount.

## Business Recommendations

1. **Prioritize the Clothing category:** Clothing had the highest total quantity sold. The business can monitor inventory levels and ensure popular items remain available.
2. **Promote Beauty products:** Beauty had the highest average transaction amount. Bundles or premium product promotions may help increase category revenue.
3. **Target the 46–55 age group:** This group generated the highest revenue. Tailored offers and relevant product recommendations may improve customer engagement.
4. **Use gender-neutral marketing insights:** Revenue from female and male customers was relatively close. Marketing campaigns should be evaluated using actual purchase behavior rather than assumptions.
5. **Monitor sales trends carefully:** Review monthly and quarterly performance regularly, while excluding incomplete periods from full-period comparisons.

## Limitation

The dataset contains product categories but does not include individual product names. Therefore, a Top 10 individual products analysis could not be performed.
```
