```python
## 2. Initial Data Quality Inspection

Initial inspection was performed using:
- `df.shape`
- `df.info()`
- `df.isnull().sum()`
- `df.duplicated().sum()`
```

```python
### Findings
- Total records: 20,790
- Total columns: 11
- Missing values: 208, all in `total_bedrooms`
- Duplicate rows: 0
- Incorrect data types: 2 columns (`total_rooms` and `sale_date`)
```

```python
## 3. Missing Data Handling

The `total_bedrooms` column contained 208 missing values.

**Treatment:** Missing values were filled using median imputation.

**Reason:** Median is less sensitive to extreme values than the mean and is suitable for numerical data with potential outliers.

Additionally, 40 invalid negative values were identified in `housing_median_age`. These were converted into missing values and filled using the column median of 29.0.
```

```python
## 4. Duplicate Removal

Duplicate records were checked using `duplicated()`.

**Finding:** No duplicate rows were found.

Therefore, no records were removed during duplicate handling.
```

```python
## 5. Data Standardisation

### total_rooms
The column contained inconsistent text formatting, including values with the word "rooms".

**Treatment:** Non-numeric text was removed and values were converted into `float64`.

### ocean_proximity
The column contained inconsistent categorical formatting, including uppercase, lowercase, and extra spaces.

**Treatment:**
- Removed leading and trailing spaces.
- Converted all category values to uppercase.

The final column contained five standardised categories:
- `<1H OCEAN`
- `INLAND`
- `NEAR OCEAN`
- `NEAR BAY`
- `ISLAND`

### sale_date
The column was originally stored as an object data type.

**Treatment:** Converted it into `datetime64[ns]` format for correct date handling.
```

```python
## 6. Outlier Detection and Treatment

The Interquartile Range (IQR) method was used to identify outliers in numerical columns.

The following formula was applied:

- IQR = Q3 − Q1
- Lower Bound = Q1 − 1.5 × IQR
- Upper Bound = Q3 + 1.5 × IQR

### Findings

Outliers were identified in:
- housing_median_age: 40
- total_rooms: 1,294
- total_bedrooms: 1,315
- population: 1,217
- households: 1,227
- median_income: 680
- median_house_value: 1,074

Longitude and latitude had no IQR outliers.

### Treatment Decisions

- Negative values in `housing_median_age` were treated as invalid, converted to missing values, and imputed using the median.
- Other extreme numerical observations were retained because they could represent genuine housing and population characteristics.
- No blanket deletion or capping of outliers was performed.
```
